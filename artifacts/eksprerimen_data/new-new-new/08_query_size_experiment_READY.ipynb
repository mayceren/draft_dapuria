# Reranking, Hybrid Retrieval, dan Guardrail

Lanjutan dari `02_embedding_indexing.ipynb`. Notebook ini dibuat standalone (menyalin ulang
parser dari notebook 01) supaya bisa dijalankan tanpa harus run notebook 01 dulu di kernel yang sama.

Isi notebook:
1. Setup model & koneksi ChromaDB
2. Parser bahan untuk input user
3. Skor efisiensi bahan (coverage_recipe + coverage_user, digabung pakai harmonic mean)
4. Hybrid retrieval (SBERT + keyword fallback, dengan tie-breaker efisiensi)
5. Reranking (Weighted Linear Combination)
6. Guardrail (cegah Llama3 "mengarang" resep) -- threshold T_jumlah, T_kategori, T_relevansi
7. Grid search bobot final (w1, w2)
8. Spot-check kualitatif dengan bobot final

> **Versi evaluasi V2:** Cell pencocokan bahan dan evaluasi ground truth sudah diperbarui untuk `ground_truth_query_v2.csv`, target berbasis `corpus_id_target`, serta Candidate Hit pada candidate pool.


> **Eksperimen retrieval relevance-set:** membandingkan 20+10, 30+20, dan 50+20 dengan WLC tetap 0.4/0.6.


> **WLC validation grid search:** notebook ini hanya memakai `ground_truth_relevance_validation.csv` (42 query). Test set tidak disentuh.


> **Ablation validation:** menambahkan semantic-only dan efficiency-only baseline, lalu mengekspor ranking failure. Test set tidak digunakan.


> **Query representation experiment:** membandingkan raw query, parsed query, dan parsed+clean query pada validation set saja. Test set belum digunakan.


> **Query-size experiment:** membandingkan BASE_V2, TOP3, TOP4, TOP5, dan MAIN_PLUS_SUPPORT pada validation set saja. Semua query dibentuk hanya dari ingredients_raw hasil scraping; target corpus tidak digunakan dalam pemilihan bahan.

## 3. Skor Efisiensi Bahan

Definisi:
- **coverage_recipe** = bahan resep yang cocok / total bahan resep -- makin tinggi, makin sedikit bahan tambahan yang perlu dibeli
- **coverage_user** = bahan user yang cocok / total bahan yang diketik user -- makin tinggi, makin banyak bahan user yang benar-benar terpakai
- **skor_efisiensi_bahan** = harmonic mean dari keduanya

**PENTING (hasil debugging panjang):** kalau cuma pakai coverage_recipe saja, resep pendek yang cuma nyerempet 1 bahan bisa dapat skor sama tingginya dengan resep yang benar-benar memanfaatkan SEMUA bahan user, cuma karena total bahannya sedikit. Harmonic mean mencegah ini -- resep hanya dapat skor tinggi kalau KEDUA coverage sama-sama tinggi.

```python
def ingredient_matches(u: str, r: str) -> bool:
    """Pencocokan bahan berbasis word/phrase boundary.

    Perbaikan dari versi lama:
    - tidak lagi memakai `u in r or r in u`
    - mencegah false match seperti "teri" terhadap "terigu"
    - tetap mengizinkan phrase yang benar-benar berdiri sebagai kata/frasa
      di dalam nama bahan yang lebih panjang.
    """
    u, r = u.lower().strip(), r.lower().strip()
    if not u or not r:
        return False
    if u == r:
        return True

    pattern_u = rf"(?<!\w){re.escape(u)}(?!\w)"
    pattern_r = rf"(?<!\w){re.escape(r)}(?!\w)"
    return bool(re.search(pattern_u, r)) or bool(re.search(pattern_r, u))


def coverage_stats(user_ingredients: list, recipe_ingredients: list):
    """Return: (coverage_recipe, coverage_user, matched_items, missing_items)"""
    if not recipe_ingredients:
        return 0.0, 0.0, [], []

    matched_recipe_items, matched_user_items = [], set()

    for r_name in recipe_ingredients:
        hit = False
        for u_name in user_ingredients:
            if ingredient_matches(u_name, r_name):
                hit = True
                matched_user_items.add(u_name)

        if hit:
            matched_recipe_items.append(r_name)

    missing_items = [r for r in recipe_ingredients if r not in matched_recipe_items]

    coverage_recipe = len(matched_recipe_items) / len(recipe_ingredients)
    coverage_user = (
        len(matched_user_items) / len(user_ingredients) if user_ingredients else 0.0
    )

    return coverage_recipe, coverage_user, matched_recipe_items, missing_items


def harmonic_efficiency(cr: float, cu: float) -> float:
    if cr + cu == 0:
        return 0.0
    return (2 * cr * cu) / (cr + cu)


def fbeta_efficiency(cr: float, cu: float, beta: float = 2.0) -> float:
    """Alternatif harmonic mean berbobot -- beta>1 menekankan coverage_user lebih penting.
    Disimpan sebagai opsi eksperimen, tidak dipakai di formula final."""
    denom = (beta**2 * cr) + cu
    if denom == 0:
        return 0.0
    return (1 + beta**2) * (cr * cu) / denom


# sanity check
assert ingredient_matches("teri", "terigu") is False
assert ingredient_matches("ayam", "dada ayam") is True

cr, cu, matched, missing = coverage_stats(
    ["telur", "dada ayam", "bawang putih"],
    ["dada ayam", "telur", "garam", "merica", "kaldu jamur", "bawang putih"],
)
print(
    f"coverage_recipe={cr:.3f}, coverage_user={cu:.3f}, "
    f"efisiensi={harmonic_efficiency(cr, cu):.3f}"
)
print(f"matched: {matched}, missing: {missing}")
```

## 4. Hybrid Retrieval (SBERT + Keyword Fallback)

**Kenapa perlu hybrid:** pengujian awal menemukan bahwa retrieval SBERT murni pada query pendek
(2-3 bahan) bisa gagal total memasukkan resep yang relevan ke top-K, karena embedding cenderung
mengelompokkan berdasarkan pola topik dominan (mis. semua hasil "Telur Dadar" walau user
menyebut "dada ayam" juga). Solusinya: gabungkan hasil SBERT dengan pencarian kata kunci
langsung ke seluruh koleksi.

**PENTING (bug yang pernah ditemukan):** `keyword_candidates` HARUS diurutkan dengan
tie-breaker `coverage_recipe`, bukan cuma `matched_count` saja -- kalau tidak, banyak resep
yang seri jumlah match-nya (puluhan resep bisa seri), dan resep yang lebih efisien bisa
kepotong dari batas `max_keyword_extra` hanya karena urutan asli di dataset, bukan karena
kualitasnya.

```python
def retrieve_candidates(user_text: str, top_k: int = 20) -> list:
    query_embedding = model.encode(
        [add_query_prefix(user_text)], normalize_embeddings=True, convert_to_numpy=True
    )
    results = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=top_k,
        include=["documents", "metadatas", "distances"],
    )
    candidates = []
    for i in range(len(results["ids"][0])):
        meta = results["metadatas"][0][i]
        candidates.append(
            {
                "id": results["ids"][0][i],
                "title": meta["title"],
                "ingredient_names": json.loads(meta["ingredient_names"]),
                "sbert_score": 1 - results["distances"][0][i],
            }
        )
    return candidates


_ALL_DOCS_CACHE = None


def get_all_documents():
    """Cache seluruh koleksi di memori sekali saja (5000 dokumen, ringan)."""
    global _ALL_DOCS_CACHE
    if _ALL_DOCS_CACHE is None:
        raw = collection.get(include=["metadatas", "documents", "embeddings"])
        _ALL_DOCS_CACHE = list(
            zip(raw["ids"], raw["metadatas"], raw["documents"], raw["embeddings"])
        )
    return _ALL_DOCS_CACHE


def keyword_candidates(
    user_ingredients: list, exclude_ids: set, max_candidates: int = 10
) -> list:
    """Diurutkan berdasarkan (matched_count, coverage_recipe) -- keduanya menurun.
    Tie-breaker coverage_recipe WAJIB ada supaya resep yang lebih efisien tidak
    kepotong hanya karena urutan asli di dataset."""
    scored = []
    for doc_id, meta, document, embedding in get_all_documents():
        if doc_id in exclude_ids:
            continue
        names = json.loads(meta["ingredient_names"])
        matched = sum(
            1 for u in user_ingredients if any(ingredient_matches(u, n) for n in names)
        )
        if matched > 0:
            coverage = matched / len(names) if names else 0.0
            scored.append((matched, coverage, doc_id, meta, document, embedding))

    scored.sort(key=lambda x: (-x[0], -x[1]))
    return [
        (m, doc_id, meta, document, embedding)
        for m, cov, doc_id, meta, document, embedding in scored[:max_candidates]
    ]


def retrieve_candidates_hybrid(
    user_text: str,
    user_ingredients: list,
    top_k_sbert: int = 20,
    max_keyword_extra: int = 10,
) -> list:
    """top_k_sbert=20 + max_keyword_extra=10 -> total maksimal 30 kandidat,
    sesuai proposal final sub-bab 3.1.3.6."""
    query_embedding = model.encode(
        [add_query_prefix(user_text)], normalize_embeddings=True, convert_to_numpy=True
    )[0]
    sbert_candidates = retrieve_candidates(user_text, top_k=top_k_sbert)
    existing_ids = {c["id"] for c in sbert_candidates}
    kw_extra = keyword_candidates(
        user_ingredients, exclude_ids=existing_ids, max_candidates=max_keyword_extra
    )
    for matched, doc_id, meta, document, embedding in kw_extra:
        sbert_score = float(np.dot(query_embedding, np.array(embedding)))
        sbert_candidates.append(
            {
                "id": doc_id,
                "title": meta["title"],
                "ingredient_names": json.loads(meta["ingredient_names"]),
                "sbert_score": sbert_score,
            }
        )
    return sbert_candidates
```

## 5. Reranking: Weighted Linear Combination

$$skor\_similarity = 1 - distance$$
$$skor\_akhir = w_1 \cdot skor\_similarity + w_2 \cdot skor\_efisiensi\_bahan$$

Bobot final: **w1 = 0,4** dan **w2 = 0,6** (hasil grid search 30 query + validasi kualitatif,
lihat sub-bab 7 dan 8 di bawah).

```python
W1_FINAL = 0.4
W2_FINAL = 0.6


def rerank_hybrid(
    user_text: str,
    w1: float = W1_FINAL,
    w2: float = W2_FINAL,
    top_n: int = 5,
    top_k_sbert: int = 20,
    max_keyword_extra: int = 10,
) -> tuple:
    user_ingredients = parse_user_ingredients(user_text)
    candidates = retrieve_candidates_hybrid(
        user_text, user_ingredients, top_k_sbert, max_keyword_extra
    )

    sbert_scores = np.array([c["sbert_score"] for c in candidates])
    s_min, s_max = sbert_scores.min(), sbert_scores.max()
    sbert_norm = (sbert_scores - s_min) / (s_max - s_min + 1e-9)

    for c, s_norm in zip(candidates, sbert_norm):
        cr, cu, matched, missing = coverage_stats(
            user_ingredients, c["ingredient_names"]
        )
        c["coverage_recipe"] = cr
        c["coverage_user"] = cu
        c["skor_efisiensi_bahan"] = harmonic_efficiency(cr, cu)
        c["sbert_score_norm"] = float(s_norm)
        c["matched_items"] = matched
        c["missing_items"] = missing
        c["final_score"] = w1 * float(s_norm) + w2 * c["skor_efisiensi_bahan"]

    top = sorted(candidates, key=lambda c: c["final_score"], reverse=True)[:top_n]
    return top, user_ingredients
```

```python
user_input = "2 butir telur, 100gr dada ayam, 2 siung bawang putih"
top, user_ingredients = rerank_hybrid(user_input)

print("Bahan user (setelah parsing):", user_ingredients)
print("=" * 70)
for i, c in enumerate(top):
    print(
        f"[{i + 1}] {c['title']}  (final={c['final_score']:.3f}, sbert_norm={c['sbert_score_norm']:.3f}, "
        f"cov_recipe={c['coverage_recipe']:.3f}, cov_user={c['coverage_user']:.3f}, "
        f"efisiensi={c['skor_efisiensi_bahan']:.3f})"
    )
    print(f"    bahan resep ({len(c['ingredient_names'])}): {c['ingredient_names']}")
    print(f"    -> sudah dimiliki user: {c['matched_items']}")
    print(f"    -> perlu ditambah: {c['missing_items']}")
    print()
```

```python
import re  # dibutuhkan is_main_ingredient() untuk word-boundary matching

# Daftar kata kunci bahan utama, disusun berdasarkan 7 dari 13 kelompok pangan resmi
# panganku.org (lihat Lampiran 1 proposal). Setiap kata kunci disusun secara manual
# dengan memeriksa entri satu per satu (bukan ekstraksi otomatis), termasuk penyertaan
# bagian organ secara terpisah dari nama hewannya (mis. "hati" terpisah dari "sapi").
MAIN_INGREDIENT_KEYWORDS = {
    "Serealia": {
        "beras",
        "nasi",
        "ketan",
        "jagung",
        "cantel",
        "jali",
        "jawawut",
        "jampang huma",
        "tepung beras",
        "tepung jagung",
        "tepung terigu",
        "maizena",
        "bihun",
        "mie",
        "mi",
        "misoa",
        "makaroni",
        "spaghetti",
        "roti",
    },
    "Umbi Berpati": {
        "kentang",
        "singkong",
        "ketela pohon",
        "ubi",
        "ubi jalar",
        "ubi cilembu",
        "talas",
        "gembili",
        "gadung",
        "ganyong",
        "suweg",
        "uwi",
        "sente",
        "bengkuang",
        "arrowroot",
        "sagu",
        "tapioka",
        "tepung singkong",
        "tepung kentang",
        "tepung ganyong",
        "tepung mocaf",
        "mocaf",
        "tepung arrowroot",
        "tepung ubi",
    },
    "Kacang-kacangan": {
        "kacang arab",
        "kacang bogor",
        "kacang ercis",
        "kacang gude",
        "kacang hijau",
        "kacang hitam",
        "kacang kapri",
        "kedelai",
        "kacang kedelai",
        "kacang komak",
        "kacang merah",
        "kacang mete",
        "kacang tanah",
        "kacang tolo",
        "kacang tunggak",
        "kacang panjang",
        "kenari",
        "koro",
        "lamtoro",
        "wijen",
        "tahu",
        "tempe",
        "oncom",
        "kembang tahu",
        "susu kedelai",
        "tauco",
        "tepung hunkwe",
        "hunkwe",
        "tepung jalejo",
        "tepung kacang kedelai",
        "tepung kedelai",
    },
    "Sayuran": {
        "akar tonjong",
        "bit",
        "lobak",
        "rebung",
        "umbut kelapa",
        "umbut rotan",
        "ale",
        "toge",
        "taoge",
        "taoge kacang kedelai",
        "taoge kacang tunggak",
        "andewi",
        "bayam",
        "bayam merah",
        "daun bangun-bangun",
        "daun bebuas",
        "daun belem",
        "daun bluntas",
        "daun gandaria",
        "daun gedi",
        "daun gedi besar",
        "daun gedi kecil",
        "daun gelang",
        "daun gunda",
        "daun gunda bali",
        "daun gunda serang",
        "daun jambu mete",
        "daun jampang",
        "daun jawaw seluang",
        "daun jonghe",
        "daun kacang ma",
        "daun kacang panjang",
        "daun kasbi",
        "daun singkong karet",
        "daun katuk",
        "daun kecipir",
        "daun kedondong",
        "daun kelor",
        "daun kemang",
        "daun kenikir",
        "kenikir",
        "daun kol sawi",
        "daun koro",
        "daun kubis",
        "daun kumak",
        "daun labu siam",
        "daun labu",
        "daun labu waluh",
        "daun labu kuning",
        "daun lamtoro",
        "daun leilem",
        "daun leunca",
        "daun lobak",
        "daun lompong",
        "daun lompong talas",
        "daun mangkokan",
        "daun matel ambon",
        "daun melinjo",
        "daun mengkudu",
        "daun ndusuk",
        "daun oyong",
        "daun pakis",
        "daun pakis longgiho",
        "daun pakis wambateu",
        "daun paku",
        "daun pangi",
        "daun pare",
        "daun pepaya",
        "daun pohpohan",
        "pohpohan",
        "daun semanggi",
        "semanggi",
        "daun simpur",
        "daun singkil",
        "daun singkong",
        "daun singkong ambon",
        "daun singkong ampenan",
        "daun singkong kopang",
        "daun sintrong",
        "sintrong",
        "daun talas",
        "daun tespong",
        "tespong",
        "daun ubi",
        "daun ubi kuning",
        "daun ubi merah",
        "daun ubi putih",
        "daun ubi tinta",
        "eceng",
        "genjer",
        "kalakai",
        "kangkung",
        "kangkung tondano",
        "selada",
        "selada air",
        "sawi",
        "sawi putih",
        "pecai",
        "sawi taiwan",
        "sawi tanah",
        "mostarda metan",
        "baligo",
        "buah kelor",
        "buah merah",
        "gambas",
        "oyong",
        "jagung muda",
        "jagung semi",
        "baby corn",
        "jantung pisang",
        "jotang",
        "kabau",
        "karawila",
        "kelawi",
        "kluwih",
        "ketimun",
        "timun",
        "ketimun krai",
        "ketimun madura",
        "kundur",
        "labu air",
        "labu kuning",
        "labu siam",
        "labu waluh",
        "waluh",
        "lantar",
        "leunca",
        "lumai",
        "nangka muda",
        "paria putih",
        "pare",
        "paria",
        "pepare ular",
        "pepaya muda",
        "rimbang",
        "tekokak",
        "terong",
        "terung",
        "terong asam",
        "terung belanda",
        "terung bengkulu",
        "terong hintalo",
        "terung panjang",
        "tomat",
        "tomat merah",
        "tomat muda",
        "buncis",
        "kacang mekah",
        "kacang panjang",
        "kacang ranti",
        "kapri muda",
        "kecipir muda",
        "komak",
        "koro kerupuk",
        "koro wedus",
        "caisin",
        "kool kembang",
        "kembang kol",
        "kol kembang",
        "kool merah",
        "kol merah",
        "kool putih",
        "kol putih",
        "kubis",
        "kol",
        "bunga pepaya",
        "bunga turi",
        "kembang turi",
        "uceng",
        "bunga melinjo",
        "jamur",
        "jamur encik",
        "jamur kuping",
        "jamur merang",
        "jamur tiram",
        "jamur sagu",
        "jengkol",
        "kerokot",
        "kulit melinjo",
        "melinjo",
        "petai",
        "putri malu",
        "rumput laut",
        "tebu terubuk",
        "terubuk",
        "wortel",
    },
    "Daging": {
        "sapi",
        "ayam",
        "angsa",
        "babi",
        "bebek",
        "itik",
        "belibis",
        "burung",
        "domba",
        "kambing",
        "kelinci",
        "kerbau",
        "kuda",
        "ulat sagu",
        "buaya",
        "menjangan",
        "punai",
        "rusa",
        "penyu",
        "anjing",
        "tikus",
        "daging",
        "babat",
        "jeroan",
        "daleman",
        "hati",
        "ginjal",
        "usus",
        "otak",
        "kikil",
        "buntut",
        "ampela",
        "ham",
        "dendeng",
        "kornet",
        "sosis",
    },
    "Ikan/Kerang/Udang/dll": {
        "belut",
        "belut laut",
        "cumi-cumi",
        "cumi",
        "ikan",
        "ikan bader",
        "ikan balong",
        "ikan bambangan",
        "ikan bandeng",
        "bandeng",
        "ikan banjar",
        "ikan baronang",
        "baronang",
        "ikan batung",
        "ikan baung",
        "baung",
        "ikan bawal",
        "bawal",
        "ikan belida",
        "belida",
        "ikan beunteur",
        "ikan biawan",
        "ikan bili",
        "ikan bubara",
        "ikan bulan-bulan",
        "ikan cakalang",
        "cakalang",
        "ikan daun",
        "ikan ekor kuning",
        "ekor kuning",
        "ikan gabus",
        "gabus",
        "ikan heu",
        "ikan hitam",
        "ikan hiu",
        "hiu",
        "ikan kacangan",
        "ikan kakap",
        "kakap",
        "kakap merah",
        "ikan kakatua",
        "ikan kalaban",
        "ikan kamera",
        "ikan kapar",
        "ikan kawalinya",
        "ikan keru-keru",
        "ikan kima",
        "ikan lais",
        "lais",
        "ikan layang",
        "layang",
        "ikan layur",
        "layur",
        "ikan lehoma",
        "ikan lemuru",
        "lemuru",
        "ikan lidah",
        "ikan malalugis",
        "ikan mamar merah",
        "ikan mas",
        "ikan mayong",
        "ikan mujahir",
        "mujahir",
        "ikan nasu metti",
        "ikan oci",
        "ikan kembung",
        "kembung",
        "ikan paling",
        "ikan papuyu",
        "papuyu",
        "betok",
        "ikan patin",
        "patin",
        "ikan pomo",
        "ikan puntin",
        "ikan saluang",
        "saluang",
        "ikan sarden",
        "sarden",
        "ikan selar",
        "selar",
        "ikan sepat",
        "sepat",
        "ikan sidat",
        "sidat",
        "ikan sunu",
        "sunu",
        "ikan tahuman",
        "ikan tarmon",
        "ikan telan",
        "ikan tembang",
        "ikan tempahas",
        "ikan terbang",
        "ikan teri",
        "teri",
        "ikan titang",
        "ikan tongkol",
        "tongkol",
        "ikan turi",
        "keong",
        "kepiting",
        "kerang",
        "kodok",
        "kura-kura",
        "kuro",
        "lokan",
        "rajungan",
        "rebon",
        "udang",
        "udang galah",
        "ikan asin",
        "ikan kering",
        "ikan asap",
        "ikan pindang",
        "pindang",
        "ikan sale",
        "ikan kayu",
        "peda",
        "jambal",
        "sarden kaleng",
        "sardines dalam kaleng",
        "rebon kering",
        "udang kering",
        "abon ikan",
        "tepung ikan",
        "tepung teri",
    },
    "Telur": {
        "telur",
    },
}

# gabungan seluruh kata kunci jadi satu set datar, dipakai untuk pengecekan cepat
MAIN_INGREDIENT_KEYWORDS_FLAT = set().union(*MAIN_INGREDIENT_KEYWORDS.values())


def is_main_ingredient(name: str) -> bool:
    """Cek apakah nama bahan cocok dengan salah satu kata kunci bahan utama pada
    Lampiran 1. Kata kunci dengan panjang <=5 huruf (mis. "mi", "kol", "teri", "ubi")
    dicocokkan menggunakan word-boundary (harus jadi kata utuh), BUKAN substring bebas --
    soalnya substring bebas pada kata kunci pendek terbukti (lewat testing) bisa salah
    kena kata yang tidak berkaitan sama sekali (mis. "mi"/"teri" nyempil di "misterius").
    Kata kunci yang lebih panjang (>5 huruf) tetap memakai pencocokan substring dua arah
    biasa (ingredient_matches), supaya kombinasi seperti "dada ayam" tetap cocok dengan
    kata kunci "ayam"."""
    name = name.lower().strip()
    for kw in MAIN_INGREDIENT_KEYWORDS_FLAT:
        if len(kw) <= 5:
            if re.search(r"\b" + re.escape(kw) + r"\b", name):
                return True
        else:
            if ingredient_matches(name, kw):
                return True
    return False


def classify_ingredient(name: str) -> str:
    """'bahan_utama' kalau cocok dengan salah satu kata kunci Lampiran 1,
    selain itu 'belum_terklasifikasi' (BUKAN otomatis 'bumbu' -- lihat catatan
    is_main_ingredient di atas)."""
    return "bahan_utama" if is_main_ingredient(name) else "belum_terklasifikasi"


def has_main_ingredient(user_ingredients: list) -> bool:
    """T_kategori: True kalau minimal 1 bahan pengguna cocok dengan salah satu
    kata kunci bahan utama pada Lampiran 1 (pengecekan positif/inclusion,
    bukan exclusion terhadap daftar bumbu)."""
    return any(is_main_ingredient(u) for u in user_ingredients)


# test -- kasus yang sama seperti sebelumnya, plus kasus baru untuk cek "belum terklasifikasi"
# dan kasus false-positive kata kunci pendek yang ditemukan lewat testing
print(
    has_main_ingredient(["garam", "gula", "kecap manis"])
)  # False -> tidak ada yg cocok kata kunci manapun
print(has_main_ingredient(["telur", "garam"]))  # True  -> "telur" cocok kelompok Telur
print(has_main_ingredient(["tahu", "tempe"]))  # True  -> cocok kelompok Kacang-kacangan
print(has_main_ingredient(["hati sapi"]))  # True  -> cocok "hati" & "sapi" (Daging)
print(
    has_main_ingredient(["bahan misterius xyz"])
)  # False -> belum terklasifikasi, bukan bumbu
```

```python
def is_relevant_enough(best_candidate: dict, user_ingredients: list) -> bool:
    """T_relevansi = min(2, jumlah bahan user):
    - user kasih 1 bahan -> wajib match bahan itu (T=1)
    - user kasih >=2 bahan -> wajib match minimal 2 bahan (T=2)
    Threshold tetap (mis. selalu >=2) salah menolak query pendek yang valid; threshold
    longgar (>=1) meloloskan kombinasi yang sebenarnya tidak related (lihat kasus
    'soda kue, kunyit' vs 'ayam, gula')."""
    if not best_candidate or not user_ingredients:
        return False
    T_relevansi = min(2, len(user_ingredients))
    return len(best_candidate.get("matched_items", [])) >= T_relevansi


def handle_user_input(user_text: str, top_n: int = 5) -> dict:
    """Pipeline utuh dgn guardrail. Return dict dgn 'status' yg menentukan respons apa
    yang boleh disampaikan Llama3 -- Llama3 TIDAK pernah membuat konten resep sendiri."""
    user_ingredients = parse_user_ingredients(user_text)

    if not user_ingredients:
        return {
            "status": "no_ingredients",
            "template": "Sepertinya belum ada bahan yang terdeteksi dari pesanmu. Boleh sebutkan bahan yang kamu punya?",
        }

    if not has_main_ingredient(user_ingredients):
        return {
            "status": "bumbu_only",
            "user_ingredients": user_ingredients,
            "template": (
                f"Bahan yang kamu sebutkan ({', '.join(user_ingredients)}) sepertinya baru bumbu/pelengkap. "
                f"Ada bahan utama juga nggak, misalnya ayam, telur, tahu, atau sayuran?"
            ),
        }

    results, _ = rerank_hybrid(user_text, top_n=top_n)

    if not is_relevant_enough(results[0] if results else None, user_ingredients):
        return {
            "status": "no_match",
            "user_ingredients": user_ingredients,
            "template": (
                f"Maaf, sistem belum menemukan resep yang cukup cocok untuk kombinasi bahan "
                f"({', '.join(user_ingredients)}). Coba sebutkan bahan lain atau kombinasi berbeda ya."
            ),
        }

    return {
        "status": "ok",
        "user_ingredients": user_ingredients,
        "results": results,
    }


# test ketiga kasus
for test_input in [
    "garam, gula, kecap manis",
    "soda kue, kunyit",
    "2 butir telur, 100gr dada ayam, 2 siung bawang putih",
]:
    print(f"Input: {test_input!r}")
    result = handle_user_input(test_input)
    print(f"  status: {result['status']}")
    if result["status"] != "ok":
        print(f"  template: {result['template']}")
    else:
        print(f"  top-1: {result['results'][0]['title']}")
    print()
```

```python
def evaluate_weight(w1: float, w2: float, top_n: int = 5) -> dict:
    """Rata-rata IAR@top_n (=coverage_recipe) dan SBERT@top_n untuk kombinasi bobot tertentu.
    PENTING: skor_efisiensi_bahan pakai harmonic_efficiency (coverage_recipe, coverage_user),
    bukan coverage_recipe saja -- lihat sub-bab 3."""
    iars, sberts = [], []
    for q in test_queries_full:
        user_ingredients = parse_user_ingredients(q)
        candidates = retrieve_candidates_hybrid(
            q, user_ingredients, top_k_sbert=20, max_keyword_extra=10
        )
        sbert_scores = np.array([c["sbert_score"] for c in candidates])
        s_min, s_max = sbert_scores.min(), sbert_scores.max()
        sbert_norm = (sbert_scores - s_min) / (s_max - s_min + 1e-9)
        for c, s_norm in zip(candidates, sbert_norm):
            cr, cu, matched, missing = coverage_stats(
                user_ingredients, c["ingredient_names"]
            )
            c["coverage_recipe"] = cr
            c["skor_efisiensi_bahan"] = harmonic_efficiency(cr, cu)
            c["sbert_score_norm"] = float(s_norm)
            c["final_score"] = w1 * float(s_norm) + w2 * c["skor_efisiensi_bahan"]
        top = sorted(candidates, key=lambda c: c["final_score"], reverse=True)[:top_n]
        iars.append(np.mean([c["coverage_recipe"] for c in top]))
        sberts.append(np.mean([c["sbert_score_norm"] for c in top]))
    return {"mean_iar": float(np.mean(iars)), "mean_sbert": float(np.mean(sberts))}


print(f"{'w1':>6} {'w2':>6} {'mean_CoverageRecipe@5':>12} {'mean_SBERT@5':>14}")
results_grid = []
for w1 in np.arange(0.1, 1.0, 0.1):
    w2 = round(1 - w1, 1)
    w1 = round(w1, 1)
    metrics = evaluate_weight(w1, w2)
    results_grid.append((w1, w2, metrics["mean_iar"], metrics["mean_sbert"]))
    print(
        f"{w1:>6} {w2:>6} {metrics['mean_iar']:>12.4f} {metrics['mean_sbert']:>14.4f}"
    )

print(
    "\nCatatan: setelah retrieval diubah menjadi top_k_sbert=20 + max_keyword_extra=10"
)
print(
    "(total maksimal 30 kandidat, sesuai proposal final 3.1.3.6), bandingkan tabel di atas"
)
print(
    "dengan Tabel 3.2 pada proposal (yang dihasilkan dari pengaturan retrieval lama, 30+20=50)."
)
print(
    "Kalau pola dan kesimpulan (elbow point sekitar w1=0.6, w1=0.4 lebih aman secara kualitatif)"
)
print(
    "masih konsisten, cukup dicatat sebagai verifikasi ulang. Kalau berbeda signifikan,"
)
print(
    "Tabel 3.2 dan Lampiran terkait grid search perlu diperbarui dengan angka baru ini."
)
```

```python
# ============================================================
# QUERY SIZE / INFORMATIVENESS EXPERIMENT - VALIDATION SET ONLY
#
# Tujuan:
# Membandingkan query bahan dengan jumlah informasi berbeda,
# tanpa menggunakan title/ingredients dari target corpus untuk membentuk query.
#
# Mode:
#   BASE_V2          : query_text_v2 yang sekarang
#   TOP3             : 3 bahan scraped paling informatif
#   TOP4             : 4 bahan scraped paling informatif
#   TOP5             : 5 bahan scraped paling informatif
#   MAIN_PLUS_SUPPORT: semua bahan utama informatif (maks 4)
#                      + hingga 2 bahan pendukung informatif (maks total 6)
#
# Retrieval dibekukan:
#   SBERT 50 + keyword extra 20
#
# Ranking dibekukan:
#   min-max semantic + WLC 0.7 semantic / 0.3 efficiency
#
# TEST SET TIDAK DIGUNAKAN.
# ============================================================

import pandas as pd
import numpy as np
import re
import math

VAL_PATH = "ground_truth_relevance_validation.csv"
GT_FINAL_PATH = "ground_truth_v2_final.csv"

SUMMARY_OUT = "hasil_query_size_validation_ringkas.csv"
DETAIL_OUT = "hasil_query_size_validation_detail.csv"
QUERY_PREVIEW_OUT = "query_size_variants_validation.csv"

TOP_K_SBERT = 50
MAX_KEYWORD_EXTRA = 20
TOP_N = 10
W1 = 0.7
W2 = 0.3

QUERY_MODES = [
    "BASE_V2",
    "TOP3",
    "TOP4",
    "TOP5",
    "MAIN_PLUS_SUPPORT",
]


# ============================================================
# A. UTILITAS GROUND TRUTH
# ============================================================


def parse_relevant_ids(text):
    return [x.strip() for x in str(text).split(" | ") if x.strip()]


def validate_gt(gt):
    required = {
        "scraped_title",
        "scraped_url",
        "query_text_v2",
        "relevant_ids",
    }
    missing = required - set(gt.columns)
    if missing:
        raise ValueError(f"Kolom wajib validation hilang: {sorted(missing)}")

    all_ids = []
    for x in gt["relevant_ids"]:
        all_ids.extend(parse_relevant_ids(x))

    unique_ids = list(dict.fromkeys(all_ids))
    existing = collection.get(ids=unique_ids, include=[])
    existing_ids = set(existing.get("ids", []))
    missing_ids = sorted(set(unique_ids) - existing_ids)

    if missing_ids:
        raise ValueError(
            f"{len(missing_ids)} relevant ID tidak tersedia di ChromaDB. "
            f"Contoh: {missing_ids[:10]}"
        )

    print(f"Validation set OK: {len(gt)} query, {len(unique_ids)} relevant ID unik.")


# ============================================================
# B. PEMBENTUKAN QUERY HANYA DARI SCRAPED INGREDIENTS
# ============================================================

# Bahan utama / anchor yang relevan dengan domain resep.
MAIN_TERMS = {
    "beras",
    "nasi",
    "ketan",
    "jagung",
    "bihun",
    "mie",
    "mi",
    "misoa",
    "makaroni",
    "spaghetti",
    "roti",
    "kentang",
    "singkong",
    "ubi",
    "talas",
    "sagu",
    "tapioka",
    "kedelai",
    "kacang tanah",
    "kacang hijau",
    "kacang merah",
    "kacang panjang",
    "tahu",
    "tempe",
    "oncom",
    "sapi",
    "ayam",
    "bebek",
    "itik",
    "domba",
    "kambing",
    "kerbau",
    "daging",
    "babat",
    "hati",
    "usus",
    "ampela",
    "sosis",
    "kornet",
    "dendeng",
    "ikan",
    "udang",
    "cumi",
    "kepiting",
    "kerang",
    "lele",
    "nila",
    "tongkol",
    "tuna",
    "salmon",
    "bandeng",
    "gurame",
    "cakalang",
    "teri",
    "kakap",
    "patin",
    "kembung",
    "tenggiri",
    "bawal",
    "mujair",
    "sarden",
    "kerapu",
    "manyung",
    "peda",
    "senangin",
    "telur",
    "bayam",
    "kangkung",
    "wortel",
    "kol",
    "sawi",
    "buncis",
    "terong",
    "tomat",
    "labu",
    "jamur",
    "selada",
    "toge",
    "taoge",
}

PROTEIN_TERMS = {
    "sapi",
    "ayam",
    "bebek",
    "itik",
    "domba",
    "kambing",
    "kerbau",
    "ikan",
    "udang",
    "cumi",
    "kepiting",
    "kerang",
    "lele",
    "nila",
    "tongkol",
    "tuna",
    "salmon",
    "bandeng",
    "gurame",
    "cakalang",
    "teri",
    "kakap",
    "patin",
    "kembung",
    "tenggiri",
    "bawal",
    "mujair",
    "sarden",
    "kerapu",
    "manyung",
    "peda",
    "senangin",
}

SYNONYMS = {
    "toge": "taoge",
    "cumi-cumi": "cumi",
    "sotong": "cumi",
    "kakap merah": "kakap",
    "kakap putih": "kakap",
    "dada ayam": "ayam",
    "paha ayam": "ayam",
    "ayam kampung": "ayam",
    "daging sapi": "sapi",
    "daging ayam": "ayam",
}

# Hanya bahan yang benar-benar sangat generik yang dipenalti kuat.
VERY_GENERIC = {
    "air",
    "garam",
    "gula",
    "gula pasir",
    "minyak",
    "minyak goreng",
    "merica",
    "lada",
    "penyedap",
    "msg",
    "kaldu bubuk",
}

# Bahan umum tetapi masih bisa menjadi sinyal hidangan.
COMMON_SUPPORT = {
    "bawang putih",
    "bawang merah",
    "kecap",
    "kecap manis",
    "cabai",
    "cabe",
    "santan",
    "kemiri",
    "serai",
    "sereh",
    "daun salam",
    "daun jeruk",
    "lengkuas",
    "jahe",
    "kunyit",
}


def normalize_text(s):
    t = str(s).lower().strip()
    for a, b in sorted(SYNONYMS.items(), key=lambda x: -len(x[0])):
        t = re.sub(rf"\b{re.escape(a)}\b", b, t)
    t = re.sub(r"\s+", " ", t)
    return t.strip()


def has_phrase(text, phrase):
    text = normalize_text(text)
    phrase = normalize_text(phrase)
    return bool(re.search(rf"(?<!\w){re.escape(phrase)}(?!\w)", text))


def clean_ingredient(raw):
    """Buang kuantitas/satuan dan note preparasi, tetapi pertahankan nama bahan."""
    s = str(raw).strip()

    # Hapus kuantitas awal: 1, 1.5, 1/2, dst.
    s = re.sub(
        r"^\s*(?:\d+\s*/\s*\d+|\d+(?:[.,]\d+)?)\s*",
        "",
        s,
    )

    # Hapus unit / quantifier awal.
    units = (
        r"(?:kg|g|gr|gram|ml|liter|ltr|sdm|sdt|"
        r"sendok makan|sendok teh|buah|bh|butir|btr|siung|"
        r"lembar|lbr|batang|btg|ikat|ekor|bungkus|sachet|"
        r"gelas|cangkir|mangkok|mangkuk|papan|piring|cup)"
    )
    s = re.sub(rf"^\s*{units}\s+", "", s, flags=re.I)

    # Ambil bagian sebelum koma sebagai nama bahan utama.
    s = re.split(r",\s*", s, maxsplit=1)[0]

    # Buang descriptor preparasi yang menempel di akhir.
    prep_patterns = [
        r"\bpotong(?:-potong)?\b.*$",
        r"\biris(?: tipis| halus| serong)?\b.*$",
        r"\bcincang(?: halus)?\b.*$",
        r"\bgeprek\b.*$",
        r"\bmemarkan\b.*$",
        r"\bhaluskan\b.*$",
        r"\bkupas\b.*$",
        r"\bbelah\b.*$",
        r"\bkocok(?: lepas)?\b.*$",
        r"\bsuwir(?:-suwir)?\b.*$",
        r"\bparut\b.*$",
        r"\bserut\b.*$",
        r"\byang sudah direbus\b.*$",
        r"\bsudah direbus\b.*$",
        r"\bdirebus\b.*$",
        r"\bdigoreng\b.*$",
        r"\bsangrai\b.*$",
    ]
    for pat in prep_patterns:
        s = re.sub(pat, "", s, flags=re.I)

    s = re.sub(
        r"\b(?:secukupnya|sesuai selera)\b.*$",
        "",
        s,
        flags=re.I,
    )

    s = re.sub(r"\([^)]*\)", " ", s)
    s = re.sub(r"\s+", " ", s)
    return s.strip(" ,.-")


def main_terms_in_item(cleaned):
    found = set()
    t = normalize_text(cleaned)

    for term in MAIN_TERMS:
        nt = normalize_text(term)
        if has_phrase(t, nt):
            # "telur ayam" / "kaldu ayam" bukan daging ayam.
            if nt == "ayam" and re.search(
                r"\b(?:telur|kaldu|rasa|penyedap)\s+ayam\b", t
            ):
                continue
            found.add(nt)

    return found


def title_anchors(title):
    t = normalize_text(title)
    found = []
    for term in sorted(MAIN_TERMS, key=lambda x: -len(x)):
        nt = normalize_text(term)
        if has_phrase(t, nt):
            if nt not in found:
                found.append(nt)
    return found


def build_ingredient_items(title, ingredients):
    """Skor setiap bahan tanpa menggunakan target corpus."""
    anchors_title = title_anchors(title)

    items = []
    seen = set()

    for raw in ingredients:
        cleaned = clean_ingredient(raw)
        norm = normalize_text(cleaned)

        if not norm or norm in seen:
            continue
        seen.add(norm)

        mains = main_terms_in_item(cleaned)
        confirmed_anchor = any(has_phrase(norm, a) for a in anchors_title)

        score = 0.0

        # Anchor yang juga benar-benar ada di ingredient list sangat penting.
        if confirmed_anchor:
            score += 30.0

        # Bahan utama mendapat prioritas.
        if mains:
            score += 15.0

        # Protein lebih diskriminatif.
        if mains & PROTEIN_TERMS:
            score += 10.0

        # Bahan terlalu generik turun kuat.
        if any(has_phrase(norm, g) for g in VERY_GENERIC):
            score -= 20.0

        # Support umum masih boleh dipakai, tetapi tidak mengalahkan main ingredient.
        if any(has_phrase(norm, g) for g in COMMON_SUPPORT):
            score += 2.0

        # Sedikit preferensi untuk nama bahan yang lebih spesifik.
        score += min(len(norm.split()), 5) * 0.25

        items.append(
            {
                "raw": raw,
                "clean": cleaned,
                "norm": norm,
                "main_terms": mains,
                "confirmed_anchor": confirmed_anchor,
                "score": score,
            }
        )

    items.sort(
        key=lambda x: (
            x["score"],
            len(x["main_terms"]),
            len(x["norm"]),
        ),
        reverse=True,
    )
    return items


def select_top_n(items, n):
    """Ambil N bahan informatif. Jika bahan non-generik kurang, fallback tetap deterministik."""
    if not items:
        return []

    preferred = [
        x for x in items if not any(has_phrase(x["norm"], g) for g in VERY_GENERIC)
    ]
    fallback = [x for x in items if x not in preferred]

    chosen = (preferred + fallback)[:n]
    return chosen


def select_main_plus_support(items):
    """Maks 4 bahan utama + maks 2 support; total maks 6."""
    main_items = [
        x
        for x in items
        if x["main_terms"] and not any(has_phrase(x["norm"], g) for g in VERY_GENERIC)
    ]

    support_items = [
        x
        for x in items
        if x not in main_items
        and not any(has_phrase(x["norm"], g) for g in VERY_GENERIC)
    ]

    chosen = main_items[:4]
    remaining_slots = 6 - len(chosen)
    if remaining_slots > 0:
        chosen.extend(support_items[: min(2, remaining_slots)])

    # Jika resep sangat sederhana, fallback sampai minimal 3 bila tersedia.
    if len(chosen) < 3:
        for x in items:
            if x not in chosen:
                chosen.append(x)
            if len(chosen) >= 3:
                break

    return chosen[:6]


def items_to_query(items):
    return ", ".join(x["clean"] for x in items if x["clean"])


# ============================================================
# C. LOAD DATA + JOIN INGREDIENTS RAW
# ============================================================

val = pd.read_csv(VAL_PATH)
gt_final = pd.read_csv(GT_FINAL_PATH)

validate_gt(val)

# Join utama berdasarkan scraped_url karena lebih stabil daripada title.
ingredients_map = (
    gt_final.drop_duplicates(subset=["scraped_url"], keep="first")
    .set_index("scraped_url")["ingredients_raw"]
    .to_dict()
)

missing_urls = [u for u in val["scraped_url"] if u not in ingredients_map]
if missing_urls:
    raise ValueError(
        f"{len(missing_urls)} validation rows tidak menemukan ingredients_raw "
        f"di ground_truth_v2_final.csv. Contoh: {missing_urls[:5]}"
    )


# ============================================================
# D. BANGUN SEMUA VARIAN QUERY
# ============================================================

variant_rows = []

for idx, row in val.iterrows():
    raw_ingredients = str(ingredients_map[row["scraped_url"]]).split(" | ")

    items = build_ingredient_items(
        row["scraped_title"],
        raw_ingredients,
    )

    top3 = select_top_n(items, 3)
    top4 = select_top_n(items, 4)
    top5 = select_top_n(items, 5)
    main_support = select_main_plus_support(items)

    variant_rows.append(
        {
            "row_index": idx,
            "scraped_title": row["scraped_title"],
            "scraped_url": row["scraped_url"],
            "BASE_V2": str(row["query_text_v2"]),
            "TOP3": items_to_query(top3),
            "TOP4": items_to_query(top4),
            "TOP5": items_to_query(top5),
            "MAIN_PLUS_SUPPORT": items_to_query(main_support),
            "n_available_clean_ingredients": len(items),
        }
    )

variants = pd.DataFrame(variant_rows)
variants.to_csv(
    QUERY_PREVIEW_OUT,
    index=False,
    encoding="utf-8-sig",
)

print("\nContoh query variants:")
print(
    variants[
        [
            "scraped_title",
            "BASE_V2",
            "TOP3",
            "TOP4",
            "TOP5",
            "MAIN_PLUS_SUPPORT",
        ]
    ]
    .head(8)
    .to_string(index=False)
)


# ============================================================
# E. RETRIEVAL + RERANKING
# ============================================================


def evaluate_query_text(query_text, relevant):
    user_ingredients = parse_user_ingredients(query_text)

    candidates = retrieve_candidates_hybrid(
        query_text,
        user_ingredients,
        top_k_sbert=TOP_K_SBERT,
        max_keyword_extra=MAX_KEYWORD_EXTRA,
    )

    if not candidates:
        return {
            "candidate_hit": 0,
            "rank": None,
            "hit5": 0,
            "hit10": 0,
            "rr": 0.0,
            "candidate_pool_size": 0,
            "top_ids": [],
            "top_titles": [],
        }

    sbert_scores = np.array(
        [float(c["sbert_score"]) for c in candidates],
        dtype=float,
    )
    s_min, s_max = sbert_scores.min(), sbert_scores.max()
    s_norm = (sbert_scores - s_min) / (s_max - s_min + 1e-9)

    ranked = []

    for c, semantic_score in zip(candidates, s_norm):
        cr, cu, matched, missing = coverage_stats(
            user_ingredients,
            c["ingredient_names"],
        )
        efficiency = harmonic_efficiency(cr, cu)
        final_score = W1 * float(semantic_score) + W2 * float(efficiency)

        ranked.append(
            {
                **c,
                "semantic_minmax": float(semantic_score),
                "efficiency": float(efficiency),
                "final_score": float(final_score),
            }
        )

    ranked.sort(
        key=lambda x: x["final_score"],
        reverse=True,
    )

    candidate_ids = [str(c["id"]) for c in candidates]
    candidate_hit = int(bool(relevant & set(candidate_ids)))

    top10 = ranked[:TOP_N]
    top_ids = [str(c["id"]) for c in top10]
    top_titles = [str(c["title"]) for c in top10]

    rank = next(
        (i + 1 for i, rid in enumerate(top_ids) if rid in relevant),
        None,
    )

    return {
        "candidate_hit": candidate_hit,
        "rank": rank,
        "hit5": int(rank is not None and rank <= 5),
        "hit10": int(rank is not None and rank <= 10),
        "rr": 0.0 if rank is None else 1.0 / rank,
        "candidate_pool_size": len(candidate_ids),
        "top_ids": top_ids,
        "top_titles": top_titles,
    }


all_detail = []
summaries = []

for mode in QUERY_MODES:
    candidate_hits = []
    hits5 = []
    hits10 = []
    rrs = []

    print("\n" + "=" * 90)
    print(f"Menjalankan mode query: {mode}")

    for idx, row in val.iterrows():
        query_text = str(
            variants.loc[variants["row_index"] == idx, mode].iloc[0]
        ).strip()

        if not query_text:
            print(f"PERINGATAN [{mode}] query kosong: {row['scraped_title']}")

        relevant = set(parse_relevant_ids(row["relevant_ids"]))

        try:
            result = evaluate_query_text(
                query_text,
                relevant,
            )
        except Exception as e:
            print(f"GAGAL [{mode}] [{idx}] {row['scraped_title']}: {e}")
            result = {
                "candidate_hit": 0,
                "rank": None,
                "hit5": 0,
                "hit10": 0,
                "rr": 0.0,
                "candidate_pool_size": 0,
                "top_ids": [],
                "top_titles": [],
            }

        candidate_hits.append(result["candidate_hit"])
        hits5.append(result["hit5"])
        hits10.append(result["hit10"])
        rrs.append(result["rr"])

        all_detail.append(
            {
                "query_mode": mode,
                "scraped_title": row["scraped_title"],
                "query_text": query_text,
                "n_query_items": len([x for x in query_text.split(",") if x.strip()]),
                "n_relevant_ids": len(relevant),
                "candidate_pool_size": result["candidate_pool_size"],
                "candidate_hit": result["candidate_hit"],
                "hit_at_5": result["hit5"],
                "hit_at_10": result["hit10"],
                "rank_first_relevant": result["rank"],
                "top10_result_ids": " | ".join(result["top_ids"]),
                "top10_result_titles": " | ".join(result["top_titles"]),
            }
        )

    summary = {
        "query_mode": mode,
        "top_k_sbert": TOP_K_SBERT,
        "max_keyword_extra": MAX_KEYWORD_EXTRA,
        "w1": W1,
        "w2": W2,
        "n_query": len(val),
        "candidate_hit_rate": float(np.mean(candidate_hits)),
        "candidate_hit_count": int(np.sum(candidate_hits)),
        "hit_at_5": float(np.mean(hits5)),
        "hit_at_5_count": int(np.sum(hits5)),
        "hit_at_10": float(np.mean(hits10)),
        "hit_at_10_count": int(np.sum(hits10)),
        "mrr": float(np.mean(rrs)),
    }

    summaries.append(summary)

    print(
        f"Candidate Hit={summary['candidate_hit_rate']:.4f} "
        f"({summary['candidate_hit_count']}/{summary['n_query']}) | "
        f"Hit@5={summary['hit_at_5']:.4f} "
        f"({summary['hit_at_5_count']}/{summary['n_query']}) | "
        f"Hit@10={summary['hit_at_10']:.4f} | "
        f"MRR={summary['mrr']:.4f}"
    )


detail_df = pd.DataFrame(all_detail)
summary_df = pd.DataFrame(summaries)

summary_df = summary_df.sort_values(
    ["hit_at_5", "mrr", "candidate_hit_rate", "hit_at_10"],
    ascending=[False, False, False, False],
).reset_index(drop=True)

detail_df.to_csv(
    DETAIL_OUT,
    index=False,
    encoding="utf-8-sig",
)

summary_df.to_csv(
    SUMMARY_OUT,
    index=False,
    encoding="utf-8-sig",
)

print("\n" + "=" * 100)
print("RINGKASAN QUERY SIZE / INFORMATIVENESS EXPERIMENT")
print("=" * 100)
print(summary_df.to_string(index=False))

best = summary_df.iloc[0]

print("\nMODE TERBAIK VALIDATION:")
print(
    f"query_mode={best['query_mode']} | "
    f"Candidate Hit={best['candidate_hit_rate']:.4f} | "
    f"Hit@5={best['hit_at_5']:.4f} | "
    f"Hit@10={best['hit_at_10']:.4f} | "
    f"MRR={best['mrr']:.4f}"
)

print("\nPENTING:")
print("- Semua query tambahan dibentuk hanya dari ingredients_raw hasil scraping.")
print(
    "- Tidak ada ingredients/title target corpus yang dipakai untuk memilih bahan query."
)
print("- Test set tetap belum disentuh.")

print("\nFILE OUTPUT:")
print(f"- {QUERY_PREVIEW_OUT}")
print(f"- {DETAIL_OUT}")
print(f"- {SUMMARY_OUT}")
```
